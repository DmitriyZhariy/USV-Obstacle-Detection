```python
from ultralytics import YOLO
```

```python
DATA_PATH = r"E:\Education\4 course 2 semester\Practice\panoptic_project\Data\LaRS_fusion\instance_yolo_dataset.yaml"
MODEL_NAME = "yolo26s-seg.pt"
```

```python
model = YOLO(MODEL_NAME)

model.train(
    data=DATA_PATH,
    epochs=20,
    imgsz=640,
    batch=4,
    device=0,
    workers=2,
    project=r"/content/drive/MyDrive/Учёба/Samsung/runs/yolo_instance_marine/yolo_base",
    name="yolo_final",
    save=True,
    plots=True,
    verbose=True
)
```

```python
model = YOLO(r"E:\Education\4 course 2 semester\Practice\panoptic_project\runs\yolo_instance_marine\yolo_base\yolo_final\weights\best.pt")
metrics = model.val(
    project=r"E:\Education\4 course 2 semester\Practice\panoptic_project\runs\yolo_instance_marine\yolo_base\yolo_final",
)

print("mAP50(B):", metrics.box.map50)
print("mAP50-95(B):", metrics.box.map)
print("mAP50(M):", metrics.seg.map50)
print("mAP50-95(M):", metrics.seg.map)
```

```python
model.predict(
    source=r"E:\Education\4 course 2 semester\Practice\panoptic_project\Data\LaRS_fusion\instance_yolo\val\images",
    project=r"E:\Education\4 course 2 semester\Practice\panoptic_project\runs\yolo_instance_marine\yolo_base\yolo_final",
    save=True,
    conf=0.25
)
```
