```python
from ultralytics import YOLO
import torch
```

```python
DATA_PATH = r"E:\Education\demo\dataset_yolo_clear\data.yaml"
MODEL_NAME = "yolo26s-seg.pt"
DEVICE = 0 if torch.cuda.is_available() else "cpu"
print(DEVICE)
```

```python
model = YOLO(MODEL_NAME)

model.train(
    data=DATA_PATH,
    epochs=100,
    imgsz=640,
    batch=4,
    device=DEVICE,
    workers=0,
    project=r"E:\Education\demo\yolo26s",
    name="runs",
    save=True,
    plots=True,
    verbose=True,
)
```
