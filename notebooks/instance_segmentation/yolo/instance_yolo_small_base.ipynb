```python
from ultralytics import YOLO
```

```python
DATA_PATH = r"E:\Educat ion\debts\3 course 2 semester\Practice\Datasets\LaRS_fusion\instance_yolo_dataset.yaml"
MODEL_NAME = "yolo26s-seg.pt"
```

```python
model = YOLO(MODEL_NAME)

model.train(
    data=DATA_PATH,
    epochs=80,
    imgsz=640,
    batch=4,
    device=0,
    workers=2,
    project=r"E:\Education\debts\3 course 2 semester\Practice\yolo",
    name="yolo26s_base",
    save=True,
    plots=True,
    verbose=True
)
```

```python
model = YOLO(r"E:\Education\debts\3 course 2 semester\Practice\yolo\yolo26s_base\weights\best.pt")
metrics = model.val(
    data=DATA_PATH,
    project=r"E:\Education\debts\3 course 2 semester\Practice\yolo\metrics",
)

print("    BOX    ")
print(f"Precision: {metrics.box.mp:.3f}")
print(f"Recall: {metrics.box.mr:.3f}")
print(f"mAP50: {metrics.box.map50:.3f}")
print(f"mAP50-95: {metrics.box.map:.3f}")

print("\n    SEG    ")
print(f"Precision: {metrics.seg.mp:.3f}")
print(f"Recall: {metrics.seg.mr:.3f}")
print(f"mAP50: {metrics.seg.map50:.3f}")
print(f"mAP50-95: {metrics.seg.map:.3f}")
```

```python
model.predict(
    source=r"E:\Education\debts\3 course 2 semester\Practice\Datasets\LaRS_fusion\instance_yolo\val\images",
    project=r"E:\Education\debts\3 course 2 semester\Practice\yolo\val",
    save=True,
    conf=0.25
)
```
