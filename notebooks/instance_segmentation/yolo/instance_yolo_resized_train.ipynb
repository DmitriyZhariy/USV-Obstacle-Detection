```python
from ultralytics import YOLO
```

```python
DATA_PATH = r"E:\Education\4 course 2 semester\Practice\panoptic_project\Data\lars_converted\yolo_dataset.yaml"
MODEL_NAME = "yolo26s-seg.pt"
```

```python
model = YOLO(MODEL_NAME)

model.train(
    data=DATA_PATH,
    epochs=50,
    imgsz=640,
    batch=8,
    device=0,
    workers=2,
    mask_ratio=4,
    overlap_mask=True,
    freeze=5,
    optimizer="AdamW",
    lr0=0.001,
    cos_lr=True,
    box=5.0,
    cls=0.5,
    mosaic=0.5,
    copy_paste=0.2,
    close_mosaic=40,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    fliplr=0.5,
    flipud=0.3,
    project=r"E:\Education\4 course 2 semester\Practice\panoptic_project\runs\yolo_instance_marine",
    name="yolo_freeze",
    save=True,
    plots=True,
    verbose=True
)
```

```python
model = YOLO(r"E:\Education\4 course 2 semester\Practice\panoptic_project\runs\yolo_instance_marine\yolo_freeze\weights\best.pt")

model.train(
    data=DATA_PATH,
    epochs=40,
    imgsz=832,
    batch=4,
    device=0,
    workers=2,
    mask_ratio=4,
    overlap_mask=True,
    optimizer="AdamW",
    lr0=0.0005,
    cos_lr=True,
    box=5.0,
    cls=0.4,
    copy_paste=0.05,
    mosaic=0.2, 
    mixup=0.1,
    close_mosaic=30,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    fliplr=0.5,
    project=r"E:\Education\4 course 2 semester\Practice\panoptic_project\runs\yolo_instance_marine",
    name="yolo_40_832",
    save=True,
    plots=True,
    verbose=True
)
```

```python
model = YOLO(r"E:\Education\4 course 2 semester\Practice\panoptic_project\runs\yolo_instance_marine\yolo_final\weights\best.pt")
metrics = model.val(
    project=r"E:\Education\4 course 2 semester\Practice\panoptic_project\runs\yolo_instance_marine",
)

print("mAP50(B):", metrics.box.map50)
print("mAP50-95(B):", metrics.box.map)
print("mAP50(M):", metrics.seg.map50)
print("mAP50-95(M):", metrics.seg.map)
```
