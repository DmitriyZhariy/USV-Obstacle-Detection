```python
from ultralytics import YOLO
```

```python
DATA_PATH = r"E:\Education\4 course 2 semester\Practice\panoptic_project\Data\LaRS_fusion\instance_yolo_dataset.yaml"
MODEL_NAME = "yolo26s-seg.pt"
```

```python
model = YOLO(MODEL_NAME)

model.train(
    data=DATA_PATH,
    epochs=30,
    imgsz=640,
    batch=6,
    device=0,
    workers=2,
    mask_ratio=2,
    overlap_mask=True,
    freeze=10,
    optimizer="AdamW",
    lr0=0.0007,
    cos_lr=True,
    box=5.0,
    cls=0.7,
    dfl=1.5,
    mosaic=0.7,
    copy_paste=0.05,
    close_mosaic=20,
    hsv_h=0.015,
    hsv_s=0.5,
    hsv_v=0.4,
    rect=False,
    fliplr=0.3,
    project=r"E:\Education\4 course 2 semester\Practice\panoptic_project\runs\yolo_instance_marine\yolo_640",
    name="yolo_freeze",
    save=True,
    plots=True,
    verbose=True
)
```

```python
model = YOLO(r"E:\Education\4 course 2 semester\Practice\panoptic_project\runs\yolo_instance_marine\yolo_640\yolo_frezze\weights\best.pt")

model.train(
    data=DATA_PATH,
    epochs=70,
    imgsz=640,
    batch=6,
    device=0,
    workers=2,
    mask_ratio=2,
    overlap_mask=True,
    freeze = 0,
    optimizer="AdamW",
    lr0=0.0003,
    cos_lr=True,
    box=5.0,
    cls=0.7,
    mosaic=0.4,
    copy_paste=0.05,
    close_mosaic=60,
    hsv_h=0.007,
    hsv_s=0.5,
    hsv_v=0.4,
    rect=False,
    fliplr=0.25,
    project=r"E:\Education\4 course 2 semester\Practice\panoptic_project\runs\yolo_instance_marine\yolo_640",
    name="yolo_final",
    save=True,
    plots=True,
    verbose=True
)
```

```python
model = YOLO(r"E:\Education\4 course 2 semester\Practice\panoptic_project\runs\yolo_instance_marine\yolo_resized\yolo_40_832\weights\best.pt")
metrics = model.val(
    data=DATA_PATH,
    project=r"E:\Education\4 course 2 semester\Practice\panoptic_project\metrics\yolo",
)

print("    BOX    ")
print(f"Precision: {metrics.box.mp:.3f}")
print(f"Recall: {metrics.box.mr:.3f}")
print(f"mAP50: {metrics.box.map50:.3f}")
print(f"mAP50-95: {metrics.box.map:.3f}")

print("\n    SEG    ")
print(f"Precision: {metrics.seg.mp:.3f}")
print(f"Recall: {metrics.seg.mr:.3f}")
print(f"mAP50: {metrics.seg.map50:.3f}")
print(f"mAP50-95: {metrics.seg.map:.3f}")
```
