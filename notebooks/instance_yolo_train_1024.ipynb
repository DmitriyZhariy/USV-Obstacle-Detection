```python
from ultralytics import YOLO
```

```python
DATA_PATH = r"/content/drive/MyDrive/Учёба/Samsung/lars_converted/yolo_dataset.yaml"
MODEL_NAME = "yolo26m-seg.pt"
```

```python
model = YOLO(MODEL_NAME)

model.train(
    data=DATA_PATH,
    epochs=15,
    imgsz=832,
    batch=4,
    device=0,
    workers=2,
    mask_ratio=1,
    overlap_mask=True,
    freeze=10,
    optimizer="AdamW",
    lr0=0.0007,
    cos_lr=True,
    box=5.0,
    cls=0.7,
    dfl=1.5,
    mosaic=0.7,
    copy_paste=0.05,
    close_mosaic=20,
    hsv_h=0.015,
    hsv_s=0.5,
    hsv_v=0.4,
    rect=False,
    fliplr=0.3,
    project=r"/content/drive/MyDrive/Учёба/Samsung/runs/yolo_instance_marine/yolo_1024",
    name="yolo_freeze",
    save=True,
    plots=True,
    verbose=True
)
```

```python
model = YOLO(r"/content/drive/MyDrive/Учёба/Samsung/runs/yolo_instance_marine/yolo_1024/yolo_freeze/weights/best.pt")

model.train(
    data=DATA_PATH,
    epochs=25,
    imgsz=1024,
    batch=8,
    device=0,
    workers=2,
    mask_ratio=2,
    overlap_mask=True,
    freeze = 0,
    optimizer="AdamW",
    lr0=0.0003,
    cos_lr=True,
    box=5.0,
    cls=0.7,
    mosaic=0.4,
    copy_paste=0.05,
    close_mosaic=60,
    hsv_h=0.007,
    hsv_s=0.5,
    hsv_v=0.4,
    rect=False,
    fliplr=0.25,
    project=r"/content/drive/MyDrive/Учёба/Samsung/runs/yolo_instance_marine/yolo_1024",
    name="yolo_final",
    save=True,
    plots=True,
    verbose=True
)
```

```python
model = YOLO(r"E:\Education\4 course 2 semester\Practice\panoptic_project\runs\yolo_instance_marine\yolo_1024\yolo_final\weights\best.pt")
metrics = model.val(
    project=r"E:\Education\4 course 2 semester\Practice\panoptic_project\runs\yolo_instance_marine\yolo_1024\yolo_final",
)

print("mAP50(B):", metrics.box.map50)
print("mAP50-95(B):", metrics.box.map)
print("mAP50(M):", metrics.seg.map50)
print("mAP50-95(M):", metrics.seg.map)
```
